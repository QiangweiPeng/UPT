```python
import scanpy as sc
import pandas as pd
```

```python
adata = sc.read_h5ad("raw/ChangYe2021.h5ad")
```

```python
adata
```

```python
adata.layers["counts"] = adata.X.copy()
```

```python
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

```python
sc.pp.highly_variable_genes(adata,n_top_genes=2000)
adata = adata[:, adata.var['highly_variable']]
```

```python
adata.write_h5ad(f"raw/ChangYe2021_hvg2000.h5ad")
```
